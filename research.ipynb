```python
from collections import deque, defaultdict
```

```python
recipes = ["bread", "sandwich"]
ingredients = [["yeast", "flour"], ["bread", "meat"]]
supplies = ["yeast", "flour", "meat"]
```

```python
graph = defaultdict(list)
indeg = {}
```

igd(A) can cook B, C
A-->B, A-->C
"A": ["B", "C"]

```python
for rcp, needed in zip(recipes, ingredients):
    indeg[rcp] = len(needed)
    for igd in needed:
        graph[igd].append(rcp)

print(graph)
print(indeg)
```

```python
queue = deque(supplies)
result = []
queue
```

```python
while queue:
    item = queue.popleft()
    
    for rcp in graph[item]:
        indeg[rcp] -= 1
        if indeg[rcp] == 0:
            result.append(rcp)
            queue.append(rcp)
```
