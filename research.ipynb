```python
from collections import deque
```

```python
isConnected = [[1,1,0],[1,1,0],[0,0,1]]
isConnected
```

```python
n = len(isConnected)
isVisited = n * [False]
numCity = 0
for c in range(n):
    if isVisited[c]:
        continue
    numCity += 1
    q = deque([c])
    while q:
        curCity = q.popleft()
        isVisited[c] = True
        for i in range(n):
            if isConnected[c][i] == 1 and not isVisited[i]:
                isVisited[i] = True
                q.append(i)

numCity

```
