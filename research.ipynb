```python
from collections import deque
```

```python
grid = [
    [2, 1, 1],
    [1, 1, 0],
    [0, 1, 1]
]
```

```python
m, n = len(grid), len(grid[0])
```

```python
fresh = 0
r_q = deque()

for r in range(m):
    for c in range(n):
        if grid[r][c] == 2:
            r_q.append((r, c))
        elif grid[r][c] == 1:
            fresh += 1

print(fresh) # If fresh == 0 then return 0
print(r_q)
```

```python
directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]
min = 0

while r_q and fresh > 0:

    for _ in range(len(r_q)):
        rx, ry = r_q.popleft()

        for dx, dy in directions:
            nx, ny = rx + dx, ry + dy
            if 0 <= nx < m and 0 <= ny < n and grid[nx][ny] == 1:
                grid[nx][ny] = 2
                fresh -= 1
                r_q.append((nx, ny))

    min += 1
```
