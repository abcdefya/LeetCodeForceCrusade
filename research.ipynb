```python
from collections import deque
```

```python
directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]
image = [[1,1,1],[1,1,0],[1,0,1]]
sr, sc = 1, 1
color = 2
m, n = len(image), len(image[0])
org = image[sr][sc]
org
```

```python
def dfs(x, y):
    image[x][y] = color
    for dx, dy in directions:
        nx, ny = x + dx, y + dy
        if 0 <= nx < m and 0 <= ny < n and image[nx][ny] == org:
            dfs(nx, ny)
        # image[nx][ny] = color

    return image
```

```python
image = dfs(sr, sc)
```
